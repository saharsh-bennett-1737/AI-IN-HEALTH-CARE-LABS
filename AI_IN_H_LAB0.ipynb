```python
import numpy as np

print("--- NumPy Operations ---")
data_array = np.array([120, 135, 118, 142, 125])
print("Mean value:", np.mean(data_array))

import pandas as pd

print("\n--- Pandas DataFrame ---")
data = {'Patient_ID': [101, 102, 103], 'Age': [45, 34, 29], 'Heart_Rate': [72, 85, 68]}
df = pd.DataFrame(data)
print(df.head())

print("\n--- File Operations ---")
df.to_csv('healthcare_lab0.csv', index=False)
print("File 'healthcare_lab0.csv' written successfully.")

df_read = pd.read_csv('healthcare_lab0.csv')
print("Data read from file:")
print(df_read)

import matplotlib.pyplot as plt

print("\n--- Matplotlib Visualization ---")
df['Heart_Rate'].plot(kind='bar', color='steelblue')
plt.title('Patient Heart Rate')
plt.xlabel('Patient Index')
plt.ylabel('Heart Rate (bpm)')
plt.show()
```
